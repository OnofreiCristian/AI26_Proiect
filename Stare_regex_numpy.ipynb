Parte de creare matrici folosind regex și numpy

```python
import re
import numpy as np
Dimensiune=input("Ce dimensiune are matricea? ")
dim=int(re.match(r'(\d+)',Dimensiune)[0])
Matrix=input("Introduce-ti informatii despre matrice:")
x=re.findall(r'\d+',Matrix)
lungime_fiecare_dimensiune=[]
for i in x[:dim]:
  lungime_fiecare_dimensiune.append(int(i))
Stare=np.zeros(lungime_fiecare_dimensiune)
print(x)
print(Stare)
```
